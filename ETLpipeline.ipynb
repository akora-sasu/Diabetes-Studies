**•	Example:** Implement an **ETL pipeline** that reads data from a file, processes it (e.g., cleans or transforms), and writes it to a database or another file format.

```python
import pandas as pd
import sqlite3
```

```python
def extract_data(file_path='/content/GhanaStructuralChange.csv'):
    try:     #It checks for errors when reading the file to ensure the file exists and is in the correct format
        data = pd.read_csv(file_path)
        print("Data successfully loaded.")
        return data
    except Exception as e:
        print(f"Error reading the file: {e}")
        return None
```

```python
# # Step 2: Transform - Clean and process the data
def transform_data(data):
     if data is not None:
        ## Example transformations
         # Drop rows with missing values
#         data.dropna(inplace=True)

        ## Remove duplicates
#         data.drop_duplicates(inplace=True)

        ## Rename columns if necessary to be lowercase with underscores, making them SQL-friendly.
        data.columns = [col.strip().lower().replace(" ", "_").replace(",", "") for col in data.columns]
        # Rename column
        data.rename(columns={'agriculture_forestry_and_fishing_value_added_(%_of_gdp)':'agric_value_added', 'agricultural_irrigated_land_(%_of_total_agricultural_land)':'percent_irrigated_land'}, inplace=True)
        data.rename(columns={'terms_of_agric_trade':'terms_agric_trade', 'agriculture_deflator':'agriculture_deflator'}, inplace=True)

        print("Data transformation complete.")
        return data
     else:
        print("No data to transform.")
        return None
```

```python
# Step 3: Load - Write the cleaned data to a database
def load_data(data, db_name, table_name):
    if data is not None:
        try:
            # Connect to the SQLite database (or create it if it doesn't exist)
            conn = sqlite3.connect(db_name)
            # Load data into a SQL table
            data.to_sql(table_name, conn, if_exists="replace", index=False)
            conn.close()
            print(f"Data successfully loaded into '{table_name}' table in '{db_name}' database.")
        except Exception as e:
            print(f"Error loading data into database: {e}")
    else:
        print("No data to load.")
```

```python
# Run the ETL pipeline
def run_etl_pipeline(file_path, db_name, table_name):
    data = extract_data(file_path)
    transformed_data = transform_data(data)
    load_data(transformed_data, db_name, table_name)
```

```python
# File path for the CSV file and SQLite database parameters
csv_file_path = "/content/GhanaStructuralChange.csv"
database_name = "C:\\Users\\felix\\OneDrive\\Desktop\\database.db"  # Assuming you want a database file
table_name = "cleaned_data"

# Execute the ETL process
run_etl_pipeline(csv_file_path, database_name, table_name)
# Execute the ETL process
run_etl_pipeline(csv_file_path, database_name, table_name)
```
